# Neural Operator Market Model Analysis

This notebook allows you to explore the training history and performance of the Neural Operator model without retraining.

```python
import json
import matplotlib.pyplot as plt
import numpy as np
import torch
import sys
import os

# Add project root to path
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))

from src.models import NeuralOperator
from src.dataset import OptionDataset
```

## 1. Load Training History

```python
history_path = '../models/training_history.json'

if os.path.exists(history_path):
    with open(history_path, 'r') as f:
        history = json.load(f)
    print("History loaded successfully.")
else:
    print("History file not found. Please run training first.")
```

## 2. Interactive Loss Plots

```python
if 'history' in locals():
    fig, axes = plt.subplots(1, 2, figsize=(15, 6))
    
    # Loss
    axes[0].plot(history['train_loss'], label='Train Loss')
    axes[0].plot(history['val_loss'], label='Val Loss')
    axes[0].set_title('Training Loss')
    axes[0].set_yscale('log')
    axes[0].legend()
    axes[0].grid(True, which='both', alpha=0.3)
    
    # MAPE
    axes[1].plot(history['train_mape'], label='Train MAPE')
    axes[1].plot(history['val_mape'], label='Val MAPE')
    axes[1].set_title('MAPE (%)')
    axes[1].set_yscale('log')
    axes[1].legend()
    axes[1].grid(True, which='both', alpha=0.3)
    
    plt.show()
```

## 3. Model Inference Demo

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = NeuralOperator(input_dim=6, latent_dim=3).to(device)

try:
    model.load_state_dict(torch.load('../models/neural_operator.pth', map_location=device))
    model.eval()
    print("Model loaded.")
except FileNotFoundError:
    print("Model weights not found.")
```
